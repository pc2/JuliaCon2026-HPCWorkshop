# Cache Bandwidth Benckmark
The following code creates 2 arrays and copies data between them.
Therefore, read and write memory accesses are performed.
You can try different number of elements to see how the bandwidth behaves.
Remember the cache hierarchy and sizes for the AMD EPYC 9655 processor:
| Cache Level | Size | Size in Bytes|
|-------------|-----:|-------------:|
| Level 1 | 48 KiB |    49152 Bytes |
| Level 2 |  1 MiB |  1048576 Bytes |
| Level 3 | 32 MiB | 33554432 Bytes |

We can measure the bandwidth, because we exactly know the amount of transferred data and can measure the runtime.

```julia
function copy_array!(a, b, num_elements, num_repeats)
    @inbounds for j in 1:num_repeats
        for i in 1:num_elements
            b[i] = a[i]
        end
    end
end

function memory_bandwidth(num_elements=2048, num_repeats=100000)
    num_load_store = 2       # number of load and store operations
    bytes_GB       = 1e9     # Bytes per GB

    b = rand(Float64, num_elements)
    a = rand(Float64, num_elements)

    e = @elapsed copy_array!(a, b, num_elements, num_repeats)

    datasize = sizeof(Float64) * num_elements * 2
    bw       = sizeof(Float64) * num_elements * num_load_store * num_repeats / (bytes_GB * e)
    println("walltime     = $e sec\ndataset size = $datasize Bytes\nbandwidth    = $bw GB/s")
end
```

```julia
memory_bandwidth(2048)
```
